```python
import pandas as pd
```

```python
df = pd.read_csv('https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-DV0101EN-SkillsNetwork/Data%20Files/Canada.csv')

print('Data read into a pandas dataframe!')
```

```python
df.head()
```

```python
df.set_index('Country', inplace=True)
```

```python
df.head(3)
df.index.name = None
df.head(3)
```

```python
years = list(map(str, range(1980, 2014)))
years
```

```python
%matplotlib inline 

import matplotlib as mpl
import matplotlib.pyplot as plt
```

```python
year = list(map(str,range(1980,2014)))

haiti = df.loc['Haiti', years]
haiti.head()
```

```python
haiti.plot()

# line chart
```

```python
haiti.plot(kind='line')

plt.title('Immigration from Haiti')
plt.ylabel('Number of immigrants')
plt.xlabel('Years')

plt.show() # need this line to show the updates made to the figure
```

```python
haiti.index = haiti.index.map(int) 
haiti.plot(kind='line')

plt.title('Immigration from Haiti')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')

# annotate the 2010 Earthquake. 
# syntax: plt.text(x, y, label)
plt.text(2000, 6000, '2010 Earthquake') # see note below

plt.show() 
```

```python
df_CI = df.loc[["India","China"],years]
df_CI
```

```python
df_CI = df_CI.transpose()
df_CI.head()
```

```python
df_CI.index = df_CI.index.map(int)
df_CI.plot(kind='line')

plt.title('Immigrants from China and India')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')

plt.show()
```

```python
inplace = True # paramemter saves the changes to the original df_can dataframe
df.sort_values(by='Total', ascending=False, axis=0, inplace=True)
df_top5 = df.head(5)

# transpose the dataframe
df_top5 = df_top5[years].transpose() 

df_top5
```

```python
df_top5.index = df_top5.index.map(int) # let's change the index values of df_top5 to type integer for plotting
df_top5.plot(kind='line', figsize=(14, 8)) # pass a tuple (x, y) size



plt.title('Immigration Trend of Top 5 Countries')
plt.ylabel('Number of Immigrants')
plt.xlabel('Years')


plt.show()
```
